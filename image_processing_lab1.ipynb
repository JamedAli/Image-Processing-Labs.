```python
import cv2
import skimage as ski
import matplotlib.pyplot as plt

img = ski.data.camera()  # NumPy array, grayscale, replaces cv2.imread(...)

plt.imshow(img)
plt.show()
```

```python
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.cm as cm

img2 = Image.fromarray(ski.data.camera())  # PIL Image object, replaces Image.open(...)

plt.imshow(img2, cmap=cm.Greys_r)
plt.show()
```

```python
cv2.imwrite('new_image.jpg', img)
```
